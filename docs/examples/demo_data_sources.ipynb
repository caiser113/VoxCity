## 개요

VoxCity는 여러 글로벌 및 지역 소스에서 데이터를 가져올 수 있습니다.

| 카테고리 | 소스 |
|----------|---------|
| **건물** | OpenStreetMap, Overture, EUBUCCO, Open Building 2.5D, Microsoft Building Footprints |
| **지형** | DeltaDTM, FABDEM, USGS 3DEP |
| **토지 피복** | ESA WorldCover, OpenStreetMap |
| **식생** | High Resolution Canopy Height Maps, OpenStreetMap |
| **영상** | Google Earth Engine (Sentinel-2, etc.) |

```python
# 설정
from voxcity.geo.utils import get_rectangle_vertices

# 예시 지역: 일본 도쿄 마루노우치
center = (139.7671, 35.6812)  # (lon, lat)
rectangle_vertices = get_rectangle_vertices(center, 200, 200)
```

## 🏢 건물 데이터 소스

### 1. OpenStreetMap (OSM)
**범위:** 글로벌 | **비용:** 무료 | **추천 용도:** 일반적인 전 세계 사용

```python
from voxcity.download.osm import get_building_data_from_osm

# OpenStreetMap에서 건물 풋프린트 다운로드
gdf_buildings_osm = get_building_data_from_osm(rectangle_vertices)
print(f"다운로드됨: {len(gdf_buildings_osm)} 개의 건물 (OSM)")
print(f"Columns: {list(gdf_buildings_osm.columns)}")
gdf_buildings_osm.head()
```
